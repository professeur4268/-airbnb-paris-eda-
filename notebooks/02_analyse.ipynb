```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/clean/listings_clean.csv")
sns.set_theme(style="whitegrid", palette="muted")

# ── Q1 : Quels arrondissements sont les plus chers ? ──
prix_quartier = (df.groupby("neighbourhood_cleansed")["price"].median().sort_values(ascending=False).head(15))
fig, ax = plt.subplots(figsize=(10, 6))
prix_quartier.plot(kind="barh", ax=ax, color="#3C3489")
ax.set_title("Prix médian par arrondissement (Top 15)", fontsize=14)
ax.set_xlabel("Prix médian (€/nuit)")
plt.tight_layout()
plt.savefig("../outputs/q1_prix_par_quartier.png", dpi=150)
plt.show()
```

```python
# ── Q2 : Quel type de logement a les meilleures notes ? ──
fig, ax = plt.subplots(figsize=(8, 5))
sns.boxplot(data=df, x="room_type", y="review_scores_rating", ax=ax, palette="Set2")
ax.set_title("Notes selon le type de logement", fontsize=14)
ax.set_xlabel("Type de logement")
ax.set_ylabel("Note moyenne")
plt.tight_layout()
plt.savefig("../outputs/q2_notes_par_type.png", dpi=150)
plt.show()
```

```python
# ── Q3 : Prix vs Nombre de reviews (corrélation) ──
fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(df["number_of_reviews"], df["price"],
           alpha=0.2, color="#1D9E75", edgecolors="none")
ax.set_xlim(0, 500); ax.set_ylim(0, 800)
ax.set_title("Corrélation : Prix vs Nombre de reviews", fontsize=14)
ax.set_xlabel("Nombre de reviews")
ax.set_ylabel("Prix (€/nuit)")
corr = df[["price", "number_of_reviews"]].corr().iloc[0,1]
ax.set_title(f"Corrélation Prix vs Reviews (r={corr:.2f})", fontsize=14)
plt.tight_layout()
plt.savefig("../outputs/q3_correlation_prix_reviews.png", dpi=150)
plt.show()
```
